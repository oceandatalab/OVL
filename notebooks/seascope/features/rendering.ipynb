```python
import time
import SEAScope.upload
from SEAScope.lib.utils import create_rendering_config
```

```python
# Select target
# -------------
target = {'granuleLevel': False,
          'sourceId': 2,
          'collectionId': 14,
          'granuleId': 0, # set to 0 if granuleLevel is False
          'variableId': 0
         }

# Keep "granuleLevel: False" and "granuleId: 0" for now since granule-specific
# configuration has not been tested
#
# Put the mouse cursor over the name of the collection in the catalogue
# (in the sidemenu) and a tooltip will give you the values for sourceId, collectionId
# and variableId
```

```python
# Connection settings
# -------------------
port = 11155
host='127.0.0.1'
```

```python
# Retrieve current rendering configuration
# ----------------------------------------
rcfg = None
with SEAScope.upload.connect(host, port) as link:
    rcfg = SEAScope.upload.rendering_config_for(link, target)
```

```python
# Customizing rendering configuration for a single granule is not supported
# yet, make sure "granuleLevel" is set to False
rcfg["target"]["granuleLevel"] = False
```

```python
import json
print(rcfg)
```

```python
# Animation settings
# ------------------
delay = 3  # Three seconds between rendering changes
```

```python
# Apply rendering changes
# -----------------------
with SEAScope.upload.connect(host, port) as link:
    # Modify colormap
    rcfg['colormap'] = 'rainbow'
    SEAScope.upload.rendering_config(link, rcfg)
    time.sleep(delay)
    
    opacity = 1.0
    while opacity > 0.0:
        rcfg['opacity'] = opacity
        SEAScope.upload.rendering_config(link, rcfg)
        opacity -= 0.02
        time.sleep(0.02 * delay)
        
    # Modify colormap
    rcfg['colormap'] = 'grayscale'
    rcfg['opacity'] = 0.8
    SEAScope.upload.rendering_config(link, rcfg)
    time.sleep(delay)    
    
    # Change filter mode
    rcfg["filterMode"] = "NEAREST"
    SEAScope.upload.rendering_config(link, rcfg)
    time.sleep(delay)
    
    # Set uniform color
    rcfg["colormap"] = ""
    rcfg["color"] = (200, 20, 120)
    SEAScope.upload.rendering_config(link, rcfg)
    time.sleep(delay)
```
