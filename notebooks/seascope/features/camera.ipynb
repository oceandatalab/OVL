```python
import time
import SEAScope.upload
```

```python
# Animation settings
# ------------------
delay = 0.0166  # Wait 1/60s between each step
```

```python
# Connection settings
# -------------------
port = 11155
host='127.0.0.1'
```

```python
# Animation loop
# --------------
altitude = 2000000.0
with SEAScope.upload.connect(host, port) as link:
    # Select Kernel > Interrupt in order to abort animation
    
    # Set initial location
    SEAScope.upload.location(link, -45.0, 48.0)
    
    # Perform a smooth zoom (5km step)
    while altitude > 1000000.0:
        SEAScope.upload.altitude(link, altitude)
        altitude -= 5000
        time.sleep(delay)

    # Set altitude to 1000km
    altitude2 = 1000000.0
    SEAScope.upload.altitude(link, altitude2)

    # Translate the camera target by a 0.1° longitudinal step
    # from -25.0° to +25.0° while keeping a constant latitude (48°).
    lon = -45.0
    lat = 48.0    
    while lon < 5.0:
        SEAScope.upload.location(link, lon, lat)
        lon += 0.1
        time.sleep(delay)


    # Zoom out with default step
    for x in range(10):
        SEAScope.upload.zoom_out(link)
        time.sleep(delay*10)
    
    # Zoom in with default step
    for x in range(10):
        SEAScope.upload.zoom_in(link)
        time.sleep(delay*10)
        
    # Reset camera to initial position and altitude
    SEAScope.upload.reset_camera(link)
```
