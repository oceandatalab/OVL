```python
import time
import datetime
import SEAScope.upload
```

```python
# Animation settings
# ------------------
t_step = 3 * 3600  # Shift current datetime by three hours at each step
delay = 0.3  # Wait 0.3s between each step
tmin = 1448928000  # Start with 2015-12-01T00:00:00 as current datetime
tmax = 1451606400  # Stop when current datetime reaches 2016-01-01
```

```python
# Connection settings
# -------------------
port = 11155
host='127.0.0.1'
```

```python
# Animation loop
# --------------
t = tmin
with SEAScope.upload.connect(host, port) as link:
    # Select Kernel > Interrupt in order to abort animation
    while t < tmax:
        dt = datetime.datetime.fromtimestamp(t, datetime.UTC)
        SEAScope.upload.current_datetime(link, dt)
        t = t + t_step
        time.sleep(delay)
```
